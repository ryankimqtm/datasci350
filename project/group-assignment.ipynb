# DATASCI 350: group project random assignment

This notebook assigns students to groups of 4-5 for the final project.  
**Groups 01-07** were formed by students. **Groups 08-09** are randomly assigned.  
Random seed is `350` for reproducibility.

> **Note:** some emails were confirmed by students. All others follow
> the `firstname.lastname@emory.edu` pattern and **should be verified** against Canvas,
> as some students use middle names or different first names in their Emory email.

```python
import random
import pandas as pd
from IPython.display import display, Markdown

# --- Known emails (confirmed by students) ---
known_emails = {
    "Minna Lai": "minna.lai@emory.edu",
    "Tianyi Zhou": "tianyi.grace.zhou@emory.edu",
    "Sarah Shao": "kexin.shao@emory.edu",
    "Serena Gu": "serena.gu@emory.edu",
    "Stan Xu": "stan.xu@emory.edu",
    "Yunshi Zhou": "yunshi.zhou@emory.edu",
    "Oliver Xie": "lxie46@emory.edu",
    "James Teng": "yteng27@emory.edu",
    "Bonnie Bao": "bonnie.bao@emory.edu",
}

def get_email(name):
    """Return known email or generate firstname.lastname@emory.edu."""
    if name in known_emails:
        return known_emails[name]
    return ".".join(name.lower().split()) + "@emory.edu"

# --- Pre-formed groups ---
group_01 = ["Minna Lai", "Sarah Shao", "Serena Gu", "Tianyi Zhou"]
group_02 = ["Christina Dai", "Eric Wang", "Joe Xia", "Kenneth He", "Tina Piltner"]
group_03 = ["Ayushi Chowksey", "Chau Anh Nguyen", "Hank Standaert", "Jacob Hillman", "Ryana Rajesh"]
group_04 = ["Bonnie Bao", "James Teng", "Oliver Xie", "Stan Xu", "Yunshi Zhou"]
group_05 = ["Eric Chen", "Steven Hu", "Yan Tian", "Yang Lyu"]
group_06 = ["Connor Lee", "Eshaan Dani", "Ihan Do", "Jonathan Luo", "Nikhil Makker"]
group_07 = ["Anny Huang", "Chrissy Wang", "Rola Ke", "Sissi Li", "Vicki Lin"]

pre_formed = [group_01, group_02, group_03, group_04, group_05, group_06, group_07]
pre_formed_names = {name for g in pre_formed for name in g}

# --- Full student roster (alphabetical) ---
# Excludes: Danilo Freire (teacher), Alix Morales (TA), Harris Wang (grader)
all_students = [
    "Anny Huang",
    "Arush Singhania",
    "Ayushi Chowksey",
    "Bonnie Bao",
    "Chau Anh Nguyen",
    "Christina Dai",
    "Chrissy Wang",
    "Connor Lee",
    "Eric Chen",
    "Eric Wang",
    "Eshaan Dani",
    "Fei Yi",
    "Hanna Florence",
    "Hank Standaert",
    "Ihan Do",
    "Jacob Hillman",
    "James Teng",
    "Joe Xia",
    "Jonathan Luo",
    "Kenneth He",
    "Lucas Huang",
    "Minna Lai",
    "Nikhil Makker",
    "Oliver Xie",
    "Rola Ke",
    "Ryana Rajesh",
    "Saahil Mardhekar",
    "Sarah Shao",
    "Selena Lin",
    "Serena Gu",
    "Sissi Li",
    "Stan Xu",
    "Steven Hu",
    "Theo Burns",
    "Tianyi Zhou",
    "Tina Piltner",
    "Vicente Martinez",
    "Vicki Lin",
    "Yan Tian",
    "Yang Lyu",
    "Yunshi Zhou",
    "Zhenyu Zou"
]

# --- Randomly assign remaining students ---
remaining = [s for s in all_students if s not in pre_formed_names]

random.seed(350)
shuffled = remaining.copy()
random.shuffle(shuffled)

# 9 remaining: 1 group of 5 + 1 group of 4 = 2 new groups
n_new_groups = 2
base_size = len(remaining) // n_new_groups      # 4
extra = len(remaining) % n_new_groups            # 1

new_groups = []
idx = 0
for i in range(n_new_groups):
    size = base_size + (1 if i < extra else 0)
    new_groups.append(shuffled[idx:idx + size])
    idx += size

# Combine all groups
all_groups = pre_formed + new_groups

print(f"Total students: {len(all_students)}")
print(f"Pre-formed: {len(pre_formed_names)} (Groups 01-07)")
print(f"Randomly assigned: {len(remaining)} (Groups 08-09)")
print(f"Total groups: {len(all_groups)}")
```

```python
# Display all groups
n_pre = len(pre_formed)
for i, group in enumerate(all_groups, 1):
    label = "(self-formed)" if i <= n_pre else "(random)"
    display(Markdown(f"### Group {i:02d} {label}"))
    df = pd.DataFrame(
        [{"Name": s, "Email": get_email(s)} for s in sorted(group)]
    )
    df.index += 1
    display(df)

# Summary table
display(Markdown("---\n### Summary"))
rows = []
for i, group in enumerate(all_groups, 1):
    formed = "self-formed" if i <= n_pre else "random"
    for s in group:
        rows.append({"Group": f"{i:02d}", "Name": s, "Email": get_email(s), "Type": formed})
summary = pd.DataFrame(rows).sort_values("Name").reset_index(drop=True)
summary.index += 1
display(summary)
```
